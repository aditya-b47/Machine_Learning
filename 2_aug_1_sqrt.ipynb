# find the sqaure root without using ** or sqrt() or pow

```python
# sqrt(1) -> 1
# sqrt(4) -> 2
# sqrt(9) -> 3

# what would be the sqrt of 5
guess = 2.5 # It is a random guess
guess*guess # it is printing 6.25 as an output which is greater than 5 so what will do is we go a little lower with 2.25
```

```python
guess = 2.25
guess*guess # now it is printing 5.0625 as an output which is still; greater than 5 but smaller than our previous predictions so now  we            know that the number is between 2 and 2.25 so we make 2 as the low and 2.25 as the high
```

```python
# pick a guess between 2 and 2.25
lo,hi = 2, 2.25
guess = (lo+hi)/2 # with this we can find a median
print(f"guess: {guess}")
print(f"Square: {guess*guess}") # guess = 2.125 and square = 4.515625 now the square is lesser than 5 now our low is 2.125 and high is 2.25
```
