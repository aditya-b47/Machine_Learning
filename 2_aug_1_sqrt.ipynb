now we use function to do solve the problem of finding a square root

```python
def sq_root(num,hi,lo,loops):
    guess = (lo+hi)/2
    for i in range(loops):
        if  guess*guess > num:
            hi = guess
        else:
            lo = guess

        guess = (lo+hi)/2

    return guess
```

```python
ans = sq_root(5,2.5,1,100)
print(f"answer is: {ans}")
```

what if i want to improve my time and space complexity??
what i'll do is i will remove extra iterations and its space, how will i do it??
=> I will remove the concept of loops and stop the code when we get the exact square or cube root at exact points
 for example if i want to find sq root of 9 and i have taken loop range of 100 and it takes only 5 steps to reach the roots then extra 95 range will go to waste
=> to concure this problem we will apply while  

```python
def sq_root(num):
    lo,hi = 0,num
    if num==0:
      return 0
    elif 0<num<1:
      lo,hi = 0,1
    guess = (lo+hi)/2
    while True:
        if abs(guess*guess-num)<0.00001:  #abs is used for Return the absolute value of the argument.
          return guess
        elif guess*guess > num:
          hi = guess
        else:
           lo = guess
        print(f"low:{lo},high:{hi}")
        guess = (lo+hi)/2

        print(f"Guess: {guess}")
        print(f"square: {guess*guess}")
    
      
```

```python
print(sq_root(0.25))
```

finding cube root 

```python
def cube_root(num,lo,hi,loops):
    guess = (lo+hi)/2         #This low and high is the way of binary search
    for i in range(loops):
        if guess*guess*guess > num:
          hi = guess
        else:
          lo = guess

        guess = (lo+hi)/2

    return guess
```

```python
answer = cube_root(50,1,27,100)
print(f"answer is : {answer}")
```
