Exercise 1.1 — Find Minimum and Maximum
Write find_min_max(numbers) that returns (minimum, maximum).

```python
def max_min(arr):
  max = arr[0]
  min = arr[0]
  for i in arr:
    if max < i:
      max = i
    if min > i:
      min = i

  return (min,max) 

arr = [7,6,8,1,10]

print(max_min(arr))
```

Exercise 1.2 — Min-Max Normalization
You want to rescale any value to [0, 1] relative to the dataset, so the minimum maps to 0 and the maximum maps to 1.

```python
def min_max_normalize(value, data):
  minimum, maximum = max_min(data)
  normalized = (value-minimum) / (maximum-minimum)
  return normalized

min_max_normalize(10, [10, 20, 30])
```

Exercise 1.3 — Compute Mean
Write compute_mean(numbers) — average of the list. Return 0 for an empty list.

```python
def mean(num):
  total_of_num = 0
  if len(num) == 0:
    return 0
  for i in num:
    total_of_num+=i
  total_num = len(num)
  compute_mean = total_of_num/total_num
  return compute_mean

mean([2, 4, 6, 8])
```

Exercise 1.4 — Standard Deviation
Standard deviation measures how spread out values are around the mean:
Break the formula into pieces you can compute step by step. You already have compute_mean.

Write compute_sd(numbers).



```python
import math
def compute_sd(numbers):
  # step 1 : compute mean
  mean_value = mean(numbers)
    # step 2 : (i - mean)² for every i sum of squared differences
  sum_sq_diff = 0
  for i in numbers:
    sum_sq_diff += (i-mean_value)**2
      # step 3 : divide by number of values to find variance(n)
  variance = sum_sq_diff/len(numbers)
  # step 4 : sqare root of variance
  sd = math.sqrt(variance)
  return sd

compute_sd([2, 4, 4, 4, 5, 5, 7, 9]) 
```

Exercise 1.5 — Detect Outliers with Z-Score
The z-score tells you how many standard deviations a value is from the mean.

Think about it: In [10, 12, 12, 13, 12, 11, 90], the value 90 looks suspicious. How far is it from the mean? And how does that distance compare to the "typical spread" (the SD)? Write a formula for "how many SDs from the mean."

Values with 
 threshold are outliers.

Write find_outliers(nums, threshold). Reuse compute_mean and compute_sd.

```python
def find_outliers(nums, threshold):
  # calculate the mean
  mean_value = mean(nums)
  # calculate standard deviation
  sd = compute_sd(nums)
  # calculate z
  outliers = []
  for i in nums:
    z = (i-mean_value)/sd
    if abs(z) > threshold:
      outliers.append(i)
  return outliers

find_outliers([10, 12, 12, 13, 12, 11, 90], 2)
```

Exercise 1.6 — Interquartile Range (IQR)
 is the median of the lower half of the sorted data; 
 is the median of the upper half.

You know

 (median of the upper half). What single number captures the width of the "middle 50%" of the data?


Think about how to find the median of just the lower or upper half of a sorted list.

Write compute_iqr(data).

```python
def find_median(data):
  mid = len(data)//2
  if len(data)%2 == 0:
    return (data[mid-1]+data[mid])/2
  else:
    return data[mid]
```

```python
def compute_iqr(data):
  mid = len(data) // 2
  if len(data) % 2 == 0:
    lower = data[:mid]
    upper = data[mid:]
  else:
    lower = data[:mid]
    upper = data[mid+1:]
  Q1 = find_median(lower)
  Q3 = find_median(upper)
  IQR = Q3-Q1

  return IQR

compute_iqr([1, 2, 3, 4, 5, 6, 7, 8, 9])
#compute_iqr([10, 20, 30, 40, 50, 60])
```
