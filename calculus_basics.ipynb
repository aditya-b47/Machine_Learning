```python
def f(x):
  return 3*x
```

```python
h = .00001
y1 = f(1)
y2 = f(1 + h)

dy_dx = (y2-y1)/h
dy_dx
```

```python
# any python function can take any function as an argument

def diff(f , x):
  h = .000001
  y1 = f(x)
  y2 = f(x+h)
  dy_dx = (y2-y1)/h

  return dy_dx
```

```python
def myf(x):
  return 3*x

diff(myf,1)
```

```python
def sq(x):
  return x*x

diff(sq,1)
```

```python
def cube(x):
  return x*x*x

diff(cube,3)
```
