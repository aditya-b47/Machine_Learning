Exercise 4.1 — Multiply Polynomial by a Scalar

List form: [2, 0, 3, 10] * 5 = [10, 0, 15, 50]

Write multiply_polynomial(poly, num).

```python
def multiply_polynomial(poly, num):
    result = []

    for i in poly:
        result.append(i * num)

    return result

multiply_polynomial([2, 0, 3, 10], 5)
```

Exercise 4.2 — Add Two Polynomials

List form: [2, 3, 4] + [1, 5, 6] = [3, 8, 10]

Tricky part: If the polynomials have different lengths, align them from the right (the constant terms line up) and treat missing higher-power terms as 0.

(5x + 2) + (3) → [5, 2] + [3] → [5, 5]

Write add_polynomials(p1, p2).

```python
def add_polynomials(p1, p2):  # this is for equal length of polynomial
  result = []
  for i,j in zip(p1,p2):
    result.append(i+j)

  return result

add_polynomials([2, 0, 3, 10], [1, 4, 0, 6])
```

```python
def add_polynomials(p1, p2):  # this is forall length of polynomials
  result = []
  len_p1 = len(p1)  
  len_p2 = len(p2)
  diff = (len_p1 - len_p2)
  if len_p1 > len_p2:
    p2 = [0]*diff+p2
  else :
    p1 = [0]*diff+p1
  for i,j in zip(p1,p2):
    result.append(i+j)

  return result

add_polynomials([5, 2], [3])  
```

```python
# write code to calculate power 
def pow(x,n):
  # use loop and  power to build it
  if (n==x):
    return 1
  else:
    return x*pow(x,n-1)

pow(2,3)
```
