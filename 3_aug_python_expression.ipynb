# Exercise 3.3 — Distance Between Two Points (2D)
This is the Euclidean distance between the two points (x1, y1) and (x2, y2).

```python
from math import *
def dist(x1,y1,x2,y2):
  return sqrt((x2-x1)**2 + (y2-y1)**2)

# dist(0,0,3,4) # ans = 5
# dist(6,4,8,2) # ans = 2.8284271247461903
dist(1, 2, 4, 6) # ams = 5
```

# Exercise 3.5 — Extend to 3D

A point in 3D space has three coordinates: (x, y, z).

```python
def three_D(x1,y1,z1,x2,y2,z2):
  return sqrt((x2-x1)**2 + (y2-y1)**2 + (z2-z1)**2)

# three_D(0, 0, 0, 1, 1, 1) # ans = 1.7320508075688772
# three_D(2, 3, 5, 2, 3, 5) # ans = 0.0
# three_D(1, 2, 3, 4, 6, 9) # ans = 7.810249675906654
```
