# Exercise 3.2 — Write `calculate_hypotenuse`
Define a function to calculate hypoteneus of a right angle triangle

```python
import math
def hyp(base,height):
    return math.sqrt(base*base + height*height)

hyp(3,4) # ans = 5 
```
