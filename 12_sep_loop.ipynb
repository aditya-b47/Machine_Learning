```python
def print_digit_in_one_line(num): #reverse order
    while num>0:
        print(f"num:{num}")
        last_digit = num%10
        print(f"lastdigit:{last_digit}")
        num = num//10
    print(num)
```

```python
print_digit_in_one_line(12345)
```
